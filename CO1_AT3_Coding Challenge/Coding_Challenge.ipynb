```python
def double(s):
    v = "aeiou"
    g = sum(1 for i in range(len(s))
            if s[i] in v and (i == 0 or s[i-1] not in v))
    return (len(s) >= 3 and s[-3] not in v and
            s[-2] in v and s[-1] not in "aeiouwxy" and g == 1)
def generate(s, f):
    if f in ("PL", "3SG"):
        if s.endswith(("s", "x", "z", "ch", "sh")):
            return s + "es"
        if s.endswith("o") and len(s) > 1 and s[-2] not in "aeiou":
            return s + "es"
        if s.endswith("y") and len(s) > 1 and s[-2] not in "aeiou":
            return s[:-1] + "ies"
        return s + "s"
    if f == "PROG":
        if len(s) >= 3 and s.endswith("e") and s[-2] not in "aeiou":
            return s[:-1] + "ing"
        return s + (s[-1] if double(s) else "") + "ing"
    end = {"PAST": "ed", "COMP": "er", "SUP": "est"}[f]
    if s.endswith("e"):
        return s + {"ed": "d", "er": "r", "est": "st"}[end]
    if s.endswith("y") and len(s) > 1 and s[-2] not in "aeiou":
        return s[:-1] + "i" + end
    return s + (s[-1] if double(s) else "") + end
S = int(input())
lex = []
for i in range(S):
    s, pos = input().split()
    lex.append((s, pos))
Q = int(input())
words = []
for i in range(Q):
    words.append(input().strip())
results = []
for word in words:
    ans = []
    for s, pos in lex:
        if word == s:
            ans.append(s + "+" + pos)
        if pos == "N":
            features = [("PL", "PL")]
        elif pos == "V":
            features = [("3SG", "3SG"), ("PROG", "PROG"), ("PAST", "PAST")]
        else:
            features = [("COMP", "COMP"), ("SUP", "SUP")]
        for f, tag in features:
            if generate(s, f) == word:
                ans.append(s + "+" + pos + "+" + tag)
    ans = sorted(set(ans))
    results.append(", ".join(ans) if ans else "UNKNOWN")
print()
print("Output:")
for result in results:
    print(result)
```

```python
def cons(s, i):
    if s[i] in "aeiou":
        return 0
    if s[i] == "y":
        return 1 if i == 0 else 1 - cons(s, i-1)
    return 1
def measure(s):
    m = 0
    for i in range(1, len(s)):
        if cons(s, i-1) and not cons(s, i):
            m += 1
    return m
def vowel(s):
    for i in range(len(s)):
        if not cons(s, i):
            return 1
    return 0
def stem(w):
    if w.endswith("sses"):
        w = w[:-2]
    elif w.endswith("ies"):
        w = w[:-2]
    elif w.endswith("ss"):
        pass
    elif w.endswith("s"):
        w = w[:-1]
    changed = 0
    if w.endswith("eed"):
        if measure(w[:-3]) > 0:
            w = w[:-1]
    elif w.endswith("ed") and vowel(w[:-2]):
        w = w[:-2]
        changed = 1
    elif w.endswith("ing") and vowel(w[:-3]):
        w = w[:-3]
        changed = 1
    if changed:
        if w.endswith(("at", "bl", "iz")):
            w += "e"
        elif len(w) >= 2 and w[-1] == w[-2] and cons(w, len(w)-1) and w[-1] not in "lsz":
            w = w[:-1]
        elif measure(w) == 1 and len(w) >= 3 and cons(w, len(w)-1) and not cons(w, len(w)-2) and cons(w, len(w)-3):
            w += "e"
    if w.endswith("y") and vowel(w[:-1]):
        w = w[:-1] + "i"
    return w
q = int(input())
words = []
for i in range(q):
    words.append(input().strip())
results = []
for w in words:
    results.append(stem(w))
print()
print("Output:")
for w in results:
    print(w)
```
