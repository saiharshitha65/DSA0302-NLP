```python

import re

patterns = {
    "Patient ID": r'P\d{4,6}',
    "Date": r'\d{2}/\d{2}/\d{4}',
    "Contact": r'[6-9]\d{9}',
    "Age": r'\d{1,3} years'
}

for name, p in patterns.items():
    print(name, bool(re.fullmatch(p, {
        "Patient ID": "P12345",
        "Date": "15/11/2026",
        "Contact": "9876543210",
        "Age": "25 years"
    }[name])))

states = {0: {'a': 1}, 1: {'b': 2}, 2: {'b': 3}}
final = 3

def nfa(s):
    current = {0}
    for ch in s:
        current = {states[q][ch] for q in current if ch in states.get(q, {})}
    return final in current

print("NFA ab:", nfa("abb"))
print("NFA rejected:", nfa("aba"))
```

TASK 11

```python

import re
import time
from nltk.stem import PorterStemmer

ps = PorterStemmer()

concepts = {
    "fracture": ["fracture", "fractures", "fractured", "fracturing"],
    "patient": ["patient", "patients", "patient's", "patiently"],
    "infection": ["infection", "infections", "infected", "infecting"],
    "treatment": ["treatment", "treatments", "treated", "treating"],
    "diagnosis": ["diagnosis", "diagnoses", "diagnostic", "diagnostically"],
    "operation": ["operation", "operations", "operated", "operating"],
    "examination": ["examination", "examinations", "examined", "examining"],
    "medicine": ["medicine", "medicines", "medicinal", "medically"],
    "injury": ["injury", "injuries", "injured", "injuring"],
    "prescription": ["prescription", "prescriptions", "prescribed", "prescribing"]
}

fst = {form: root for root, forms in concepts.items() for form in forms}

record_types = ["Discharge", "Prescription", "Operation", "Nursing"]

filler = "clinical hospital record patient medical care health doctor ward report patient history clinical examination medical treatment".split()

records = []

for i in range(500):
    root = list(concepts)[i % 10]
    form = concepts[root][i % 4]
    text = " ".join([form] + [filler[j % len(filler)] for j in range(49)])
    records.append({
        "id": i + 1,
        "type": record_types[i % 4],
        "root": root,
        "text": text
    })

queries = []

for root, forms in concepts.items():
    for form in forms:
        queries.append((form, root))

def normalize(word, method):
    word = word.lower()

    if method == "None":
        return word

    if method == "Porter":
        return ps.stem(word)

    return fst.get(word, word)

def retrieve(query, method):
    q = normalize(query, method)
    result = []

    for record in records:
        words = re.findall(r'\b[a-z]+\b', record["text"].lower())
        if q in [normalize(w, method) for w in words]:
            result.append(record["id"])

    return result

def evaluate(actual, predicted):
    tp = len(actual & predicted)
    fp = len(predicted - actual)
    fn = len(actual - predicted)

    p = tp / (tp + fp) if tp + fp else 0
    r = tp / (tp + fn) if tp + fn else 0
    f = 2 * p * r / (p + r) if p + r else 0

    return p, r, f

methods = ["None", "Porter", "FST"]
results = {}

print("RETRIEVAL EVALUATION")

for method in methods:
    scores = []
    start = time.time()

    for query, root in queries:
        actual = {r["id"] for r in records if r["root"] == root}
        predicted = set(retrieve(query, method))
        scores.append(evaluate(actual, predicted))

    elapsed = time.time() - start

    results[method] = [
        sum(x[0] for x in scores) / len(scores),
        sum(x[1] for x in scores) / len(scores),
        sum(x[2] for x in scores) / len(scores),
        elapsed
    ]

    print("\n", method)
    print("Precision:", round(results[method][0], 3))
    print("Recall:", round(results[method][1], 3))
    print("F1 Score:", round(results[method][2], 3))
    print("Runtime:", round(elapsed, 4), "seconds")

print("\nCOVERAGE")

for method in methods:
    covered = 0
    total = 0

    for record in records:
        words = re.findall(r'\b[a-z]+\b', record["text"].lower())
        total += len(words)
        covered += sum(normalize(w, method) != w or w in fst or w in filler for w in words)

    print(method, ":", round(covered / total * 100, 2), "%")

print("\nRESOURCE REQUIREMENTS")
print("No Normalization: Low maintenance, low processing effort")
print("Porter: Low maintenance, possible over-stemming")
print("FST: High initial lexicon effort, controlled analysis")

print("\nRECOMMENDATION")
print("Use FST as the primary morphological analyser.")
print("Use Porter stemming as a fallback for unknown words.")
print("Retain original words for accurate clinical interpretation.")
```

```python

import re
import difflib
from nltk.stem import PorterStemmer

ps = PorterStemmer()

clean_records = [
    "Patient ID P10234 diagnosis fracture treatment paracetamol 500mg",
    "Patient had fever paracetamol 500mg",
    "Patient ID P10236 diagnosis fracture paracetamol 500mg"
]

noisy_records = [
    "Patient ID P1O234 diagnosis fract-\nure paracetam0l 500mg",
    "patient ku fever irundhuchu paracetamol 500mg kuduthen",
    "Patient ID P1O236 diagnosis fractu re paracetam0l 500mg kuduthen"
]

regional = {
    "ku": "to",
    "irundhuchu": "had",
    "kuduthen": "gave"
}

vocabulary = [
    "patient", "diagnosis", "fracture", "treatment",
    "paracetamol", "fever", "had", "gave", "to"
]

def original(text):
    return text.lower()

def adapt(text):
    text = re.sub(r'-\s*\n\s*', '', text)
    text = re.sub(r'\bP([0-9OIl]+)\b',
                  lambda m: "P" + m.group(1).translate(str.maketrans("OIl", "011")), text)
    text = text.replace("fractu re", "fracture")
    text = text.replace("paracetam0l", "paracetamol")

    words = text.lower().split()
    result = []

    for word in words:
        word = regional.get(word, word)
        match = difflib.get_close_matches(word, vocabulary, n=1, cutoff=0.75)
        result.append(match[0] if match else word)

    return " ".join(result)

def extract(text):
    return {
        "Patient ID": re.findall(r'\bP\d{6}\b', text, re.I),
        "Drug": re.findall(r'\b[A-Za-z]+\s*\d+\s?mg\b', text, re.I)
    }

gold = [
    {(1, "Patient ID", "P10234"), (1, "Drug", "paracetamol 500mg")},
    {(2, "Drug", "paracetamol 500mg")},
    {(3, "Patient ID", "P10236"), (3, "Drug", "paracetamol 500mg")}
]

def entity_results(records):
    found = set()

    for i, text in enumerate(records, 1):
        for category, values in extract(text).items():
            for value in values:
                found.add((i, category, value.lower().replace(" ", "")))

    return found

gold_entities = {
    (i, category, value.replace(" ", ""))
    for group in gold for i, category, value in group
}

def metrics(actual, predicted):
    tp = len(actual & predicted)
    fp = len(predicted - actual)
    fn = len(actual - predicted)

    p = tp / (tp + fp) if tp + fp else 0
    r = tp / (tp + fn) if tp + fn else 0
    f = 2*p*r/(p+r) if p+r else 0

    return round(p, 3), round(r, 3), round(f, 3)

adapted_records = [adapt(x) for x in noisy_records]

print("RECORD NORMALIZATION")

for i, text in enumerate(noisy_records):
    print("\nOriginal:", text)
    print("Adapted:", adapted_records[i])

print("\nENTITY EXTRACTION")

for i, text in enumerate(adapted_records, 1):
    print("Record", i, extract(text))

print("\nENTITY EVALUATION")

for name, records in [
    ("Clean Baseline", clean_records),
    ("Noisy Original", noisy_records),
    ("Adapted Pipeline", adapted_records)
]:
    score = metrics(gold_entities, entity_results(records))
    print(name, "Precision:", score[0],
          "Recall:", score[1], "F1:", score[2])

english = set(vocabulary) - set(regional.values())
regional_words = set(regional)

def coverage(records, adapted_mode=False):
    known = 0
    total = 0

    for text in records:
        words = re.findall(r'[a-z]+', text.lower())

        for word in words:
            total += 1
            if word in vocabulary or word in regional_words:
                known += 1

    return round(known / total * 100, 2) if total else 0

print("\nANALYSER COVERAGE")
print("Original:", coverage(noisy_records), "%")
print("Adapted:", coverage(adapted_records), "%")

queries = {
    "fracture": {1, 3},
    "fever": {2},
    "paracetamol": {1, 2, 3}
}

def search(records, query):
    return {
        i for i, text in enumerate(records, 1)
        if query in text.lower().split()
    }

def retrieval_score(records):
    scores = []

    for query, relevant in queries.items():
        predicted = search(records, query)
        scores.append(metrics(relevant, predicted))

    return tuple(round(sum(x[j] for x in scores)/len(scores), 3)
                 for j in range(3))

print("\nRETRIEVAL EVALUATION")

for name, records in [
    ("Clean Baseline", clean_records),
    ("Noisy Original", noisy_records),
    ("Adapted Pipeline", adapted_records)
]:
    p, r, f = retrieval_score(records)
    print(name, "Precision:", p, "Recall:", r, "F1:", f)

print("\nLANGUAGE IDENTIFICATION")

for word in "patient ku fever irundhuchu paracetamol kuduthen".split():
    if word in regional:
        print(word, "Regional")
    else:
        print(word, "English or unknown")

print("\nFINAL DESIGN")
print("OCR Correction")
print("Weighted/Fuzzy Matching")
print("Token-level Language Identification")
print("Regional Language Normalization")
print("Extended Morphological Lexicon")
print("Porter Stemmer Fallback")
print("Clinical Search")
```
