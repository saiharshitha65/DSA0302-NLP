```python
import re

resume = input("Enter resume details:\n")

name = re.search(r"Name:\s*(.*)", resume, re.I)
email = re.search(r"Email:\s*([\w.-]+@[\w.-]+\.\w+)", resume, re.I)
mobile = re.search(r"Mobile:\s*(\d{10})", resume, re.I)
experience = re.search(r"Experience:\s*(\d+)\s*years?", resume, re.I)

skills_list = ["Python", "Java", "SQL", "Machine Learning", "NLP"]
skills = []

for skill in skills_list:
    if re.search(r"\b" + re.escape(skill) + r"\b", resume, re.I):
        skills.append(skill)

print("\n========== CANDIDATE PROFILE ==========")
print("Name       :", name.group(1) if name else "Not Found")
print("Email      :", email.group(1) if email else "Not Found")
print("Mobile     :", mobile.group(1) if mobile else "Not Found")
print("Skills     :", ", ".join(skills) if skills else "Not Found")
print("Experience :", experience.group(1) + " years" if experience else "Not Found")

print("\n========== ELIGIBILITY ==========")

if experience and int(experience.group(1)) >= 2 and "Python" in skills:
    print("Status     : ELIGIBLE")
    print("Reason     : Minimum 2 years of experience and Python skill satisfied.")
else:
    print("Status     : NOT ELIGIBLE")
    print("Reason     : Minimum eligibility criteria not satisfied.")
```

```python
import re

products = [
    "Apple iPhone 15",
    "Samsung Galaxy S24",
    "Apple MacBook Air",
    "Dell Laptop",
    "HP Laptop",
    "Samsung Galaxy Buds",
    "Apple AirPods",
    "Sony Headphones",
    "OnePlus Nord",
    "Realme Phone"
]

keyword = input("Enter search keyword: ")
print("\n1. Exact Search")
print("2. Prefix Search")
print("3. Suffix Search")
print("4. Partial Search")

choice = input("Enter search type: ")

if choice == "1":
    pattern = r"^" + re.escape(keyword) + r"$"
elif choice == "2":
    pattern = r"^" + re.escape(keyword)
elif choice == "3":
    pattern = re.escape(keyword) + r"$"
else:
    pattern = re.escape(keyword)

matches = []

for product in products:
    if re.search(pattern, product, re.I):
        matches.append(product)

print("\n========== SEARCH RESULTS ==========")

if matches:
    for product in matches:
        print(product)
else:
    print("No matching products found.")

print("Total Matching Products:", len(matches))
```

```python
import re

reg_no = input("Enter Register Number: ")
email = input("Enter Institutional Email: ")
course = input("Enter Course Code: ")
semester = input("Enter Semester: ")
mobile = input("Enter Mobile Number: ")

valid_reg = re.fullmatch(r"22[A-Z]{2}\d{3}", reg_no)
valid_email = re.fullmatch(r"[\w.-]+@simats\.edu\.in", email)
valid_course = re.fullmatch(r"DSA\d{2}", course)
valid_semester = re.fullmatch(r"[1-8]", semester)
valid_mobile = re.fullmatch(r"[6-9]\d{9}", mobile)

print("\n========== VALIDATION RESULTS ==========")

print("Register Number :", "Valid" if valid_reg else "Invalid")
print("Email           :", "Valid" if valid_email else "Invalid")
print("Course Code     :", "Valid" if valid_course else "Invalid")
print("Semester        :", "Valid" if valid_semester else "Invalid")
print("Mobile Number   :", "Valid" if valid_mobile else "Invalid")

if valid_reg and valid_email and valid_course and valid_semester and valid_mobile:
    print("\nRegistration Status: SUCCESSFUL")
else:
    print("\nRegistration Status: UNSUCCESSFUL")
```
