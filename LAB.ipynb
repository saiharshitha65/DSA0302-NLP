```python
from nltk.tokenize import sent_tokenize, word_tokenize
from collections import Counter
text = "Python is easy to learn. Python is powerful and easy to use. I like Python."
sentences = sent_tokenize(text)
words = [w.lower() for w in word_tokenize(text) if w.isalpha()]
print("Sentences:", sentences)
print("Words:", words)
print("Number of words:", len(words))
print("Most frequent word:", Counter(words).most_common(1)[0])
```

```python
import re

text = """Call Rahul on 9876543210 or Priya on 9123456789.
Email Rahul@gmail.com and Priya@yahoo.com.
The meeting is on 25-10-2026 at Chennai."""

# Find first date
date = re.search(r'\d{2}-\d{2}-\d{4}', text)
print("First date:", date.group())

# Find all mobile numbers
mobile = re.findall(r'\b[6-9]\d{9}\b', text)
print("Mobile numbers:", mobile)

# Find all email IDs
email = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)
print("Email IDs:", email)

# Find all capitalized words
capital = re.findall(r'\b[A-Z][a-z]*\b', text)
print("Capitalized words:", capital)

# Check whether text starts with "Call"
print("Starts with Call:", bool(re.match(r'^Call', text)))

# Change date format DD-MM-YYYY to YYYY/MM/DD
text = re.sub(r'(\d{2})-(\d{2})-(\d{4})', r'\3/\2/\1', text)
print("Changed date:", text)

# Hide phone numbers
text = re.sub(r'\b[6-9]\d{9}\b', 'XXXXXXXXXX', text)
print("Hidden phone numbers:", text)
```

```python
import re

tests = {
    "a?": ["", "a", "aa"],
    "a*": ["", "a", "aaa"],
    "a+": ["a", "aaa", "b"],
    "[abc]": ["a", "b", "d"],
    ".": ["a", "ab"],
    "[^abc]": ["d", "a"],
    "^abc$": ["abc", "abcd"]
}

for pattern, strings in tests.items():
    print("\nPattern:", pattern)
    for s in strings:
        print(s, "->", "Yes" if re.fullmatch(pattern, s) else "No")
```

```python
def fsa(s):
    state = 0
    path = [0]

    for ch in s:
        if state == 0:
            state = 1 if ch == 'a' else 0
        elif state == 1:
            state = 2 if ch == 'b' else 1
        else:
            state = 1 if ch == 'a' else 0
        path.append(state)

    print(s, ":", path, "->", "Accepted" if state == 2 else "Rejected")

for s in ["ab", "aab", "abab", "abb", "ba"]:
    fsa(s)
```

```python
import re

text = "My phone number is 9876543210 and my email is abc@gmail.com."

# Search for mobile number
phone = re.search(r'\d{10}', text)

# Search for email
email = re.search(r'\w+@\w+\.\w+', text)

print("Phone number:", phone.group())
print("Email:", email.group())

# Find all numbers
numbers = re.findall(r'\d+', text)
print("Numbers:", numbers)
```

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')

text = "The children are running and playing in the gardens."

words = word_tokenize(text)
stemmer = PorterStemmer()

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
def plural(word):
    if word.endswith("s"):
        return word + "es"
    elif word.endswith("y"):
        return word[:-1] + "ies"
    else:
        return word + "s"

words = ["cat", "bus", "baby", "book"]

for word in words:
    print(word, "->", plural(word))
```
