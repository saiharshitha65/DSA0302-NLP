```python
def automaton(s):
    state = 0
    for ch in s:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == 'a':
                state = 1
            elif ch == 'b':
                state = 2
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0
    if state == 2:
        print("Accepted")
    else:
        print("Rejected")
s = input("Enter a string: ")
automaton(s)
```

```python
def plural(word):
    state = "START"
    if word.endswith("y"):
        state = "Y"
    elif word.endswith(("s", "sh", "ch", "x", "z")):
        state = "S"
    else:
        state = "NORMAL"
    if state == "Y" and len(word) > 1 and word[-2] not in "aeiou":
        return word[:-1] + "ies"
    elif state == "S":
        return word + "es"
    else:
        return word + "s"
words = ["cat", "bus", "box", "baby", "dish"]
for word in words:
    print(word, "->", plural(word))
```

```python
import nltk
from nltk.util import bigrams
from nltk.tokenize import word_tokenize
from collections import Counter
nltk.download('punkt')
nltk.download('punkt_tab')
text = "Natural language processing is interesting. Natural language processing uses machine learning."
words = word_tokenize(text.lower())
bg = list(bigrams(words))
model = Counter(bg)
print("Bigram Model:")
for pair, count in model.items():
    print(pair, ":", count)
print("\nGenerated Text:")
word = "natural"
sentence = [word]
for i in range(6):
    if word in [p[0] for p in bg]:
        next_word = next(p[1] for p in bg if p[0] == word)
        sentence.append(next_word)
        word = next_word
    else:
        break
for w in sentence:
    print(w)
```

```python
import nltk
from nltk.tag import RegexpTagger
patterns = [
    (r'^The$|^the$|^a$|^an$', 'DT'),
    (r'^is$|^am$|^are$|^was$|^were$', 'VB'),
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*ly$', 'RB'),
    (r'.*ous$|.*ful$|.*able$', 'JJ'),
    (r'.*s$', 'NNS'),
    (r'.*', 'NN')
]
tagger = RegexpTagger(patterns)
sentence = "The student is reading books quickly".split()
result = tagger.tag(sentence)
print("Rule Based POS Tagging:")
for word, tag in result:
    print(word, ":", tag)
```

EXP 10

```python
import nltk
words = "The students play football and enjoy games".split()
tagged = [(word, "NN") for word in words]
for i in range(len(tagged)):
    word, tag = tagged[i]
    if word.lower() in ["the", "and"]:
        tagged[i] = (word, "DT" if word.lower() == "the" else "CC")
    elif word.lower() in ["play", "enjoy"]:
        tagged[i] = (word, "VB")
    elif word.endswith("s"):
        tagged[i] = (word, "NNS")
print("Transformation Based Tagging:")
for word, tag in tagged:
    print(word, ":", tag)
```

```python
import nltk
from nltk import CFG
from nltk.parse import ChartParser
grammar = CFG.fromstring("""
S -> NP_SG VP_SG | NP_PL VP_PL
NP_SG -> Det N_SG
NP_PL -> Det N_PL
VP_SG -> V_SG NP_OBJ
VP_PL -> V_PL NP_OBJ
NP_OBJ -> Det N_OBJ
Det -> 'the' | 'a' | 'an'
N_SG -> 'boy' | 'girl'
N_PL -> 'boys' | 'girls'
N_OBJ -> 'apple' | 'ball'
V_SG -> 'eats' | 'likes'
V_PL -> 'eat' | 'like'
""")
parser = ChartParser(grammar)
sentence = input("Enter a sentence: ").lower().split()
trees = list(parser.parse(sentence))
if trees:
    print("Correct Subject-Verb Agreement")
    for tree in trees:
        print(tree)
else:
    print("Incorrect Subject-Verb Agreement")
```

```python
import re
def parse(expression):
    pattern = r'([A-Za-z]+)\(([^()]*)\)'
    match = re.fullmatch(pattern, expression.strip())
    if match:
        predicate = match.group(1)
        arguments = match.group(2).split(',')
        print("Valid FOPC Expression")
        print("Predicate:", predicate)
        print("Arguments:", [arg.strip() for arg in arguments])
    else:
        print("Invalid FOPC Expression")
expression = input("Enter a logical expression: ")
parse(expression)
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Natural language processing deals with human language",
    "Machine learning is used in artificial intelligence",
    "Python is a popular programming language",
    "NLP is used for text processing and analysis"
]

query = input("Enter search query: ")

vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(documents + [query])

similarity = cosine_similarity(tfidf[-1], tfidf[:-1])

print("\nDocument Ranking:")

ranking = similarity.flatten().argsort()[::-1]

for i in ranking:
    print("Document", i + 1, ":", round(similarity[0][i], 3))
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
text = "Ravi bought a car. He likes it."
sentences = nltk.sent_tokenize(text)
previous_nouns = []
for sentence in sentences:
    words = word_tokenize(sentence)
    tags = pos_tag(words)
    for word, tag in tags:
        if word.lower() in ['he', 'she', 'it', 'they']:
            if previous_nouns:
                print(word, "refers to", previous_nouns[-1])
        elif tag.startswith('NN'):
            previous_nouns.append(word)
```

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
text = input("Enter a text: ")
sentences = sent_tokenize(text)
stop_words = set(stopwords.words('english'))
scores = []
for sentence in sentences:
    words = word_tokenize(sentence.lower())
    words = [w for w in words if w.isalpha() and w not in stop_words]
    scores.append(set(words))
print("\nCoherence Analysis:")
for i in range(len(scores) - 1):
    common = scores[i].intersection(scores[i + 1])
    print("Sentence", i + 1, "and", i + 2, ":", len(common), "common words")
if len(scores) > 1:
    total = sum(len(scores[i].intersection(scores[i + 1])) for i in range(len(scores) - 1))
    print("Coherence Score:", round(total / (len(scores) - 1), 2))
else:
    print("Enter at least two sentences")
```

EXP 24

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
def dialogue_act(sentence):
    words = word_tokenize(sentence.lower())
    if sentence.endswith("?"):
        return "Question"
    elif any(word in words for word in ["hello", "hi", "hey"]):
        return "Greeting"
    elif any(word in words for word in ["thanks", "thank"]):
        return "Thanking"
    elif any(word in words for word in ["please", "kindly"]):
        return "Request"
    elif any(word in words for word in ["bye", "goodbye"]):
        return "Farewell"
    else:
        return "Statement"
sentence = input("Enter a sentence: ")
print("Dialogue Act:", dialogue_act(sentence))
```
